# Project: Expectation Decider
**Role:** Junior Data Analyst  
**Objective:** Analyze the probability patterns of student success based on historical data.

```python
import pandas as pd
import numpy as np
import scipy.stats as stats

# Generating the dataset
data = {
    'study_hours': np.random.randint(5, 15, 200),
    'attendance': np.random.randint(60, 100, 200),
    'group_discussion': ['Yes']*110 + ['No']*90,
    'previous_test_score': np.random.randint(40, 95, 200),
    'final_exam_pass': ['Pass']*90 + ['Fail']*20 + ['Pass']*30 + ['Fail']*60
}
df = pd.DataFrame(data)
df.loc[0:69, 'attendance'] = np.random.randint(81, 100, 70)
df.loc[0:69, 'study_hours'] = np.random.randint(11, 15, 70)
df.to_csv('expectation_decider_dataset.csv', index=False)
df.head()
```

### 1. Types of Events
*   **Empirical Probability:** Probability of a student passing based on this dataset.
*   **Theoretical Probability:** Probability of randomly selecting one specific student out of 200.

```python
total_students = len(df)
total_passed = len(df[df['final_exam_pass'] == 'Pass'])

empirical_prob = total_passed / total_students
theoretical_prob = 1 / total_students

print(f"Empirical Probability (Passing): {empirical_prob:.2f}")
print(f"Theoretical Probability: {theoretical_prob:.4f}")
```

### 2. Random Variable & Probability Distribution
Let random variable X = "Number of students passing" out of 3 randomly selected students.
We model this using a Binomial Distribution where n=3 and p=0.60.

```python
n = 3
p = empirical_prob
x_values = np.arange(0, n+1)
probabilities = stats.binom.pmf(x_values, n, p)

dist_table = pd.DataFrame({'X (Passes)': x_values, 'P(X)': probabilities})
print("Probability Distribution Table:\n", dist_table)
print(f"\nMean: {n * p:.2f}")
print(f"Variance: {n * p * (1 - p):.2f}")
```

```python
# 3. Contingency Table & Conditional Probability
crosstab_gd = pd.crosstab(df['group_discussion'], df['final_exam_pass'], margins=True, margins_name="Total")
print("Contingency Table:\n", crosstab_gd)

gd_and_pass = crosstab_gd.loc['Yes', 'Pass']
total_gd = crosstab_gd.loc['Yes', 'Total']

print(f"\nJoint Probability (GD AND Pass): {gd_and_pass / total_students:.4f}")
print(f"Conditional Probability (Pass | GD Yes): {gd_and_pass / total_gd:.4f}")
```


### 3. Bayes Theorem Application
Given historical data assumptions:
* P(HighAtt | Pass) = 0.70
* P(HighAtt | Fail) = 0.40
* P(HighAtt) = 0.60
* P(Pass) = 0.667 (using Law of Total Probability)

Applying Bayes Theorem:
P(Pass | HighAtt) = (0.70 * 0.667) / 0.60 = 0.7778

**Result:** A student with high attendance has a 77.78% chance of passing.
